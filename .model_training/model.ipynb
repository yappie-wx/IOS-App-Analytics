```python
import pandas as pd
```

```python
df = pd.read_csv('Labelled.csv')

df.head()
```

```python
df['intent'].value_counts()
```

```python
# selected_data = df[df['confidence'] == 'high']
selected_data = df

# Print the selected data
selected_data
```

```python
dataset = selected_data.drop(['app_version', 'last_updated', 'author', 'confidence', 'reason'], axis=1)

dataset
```

```python
dataset['text'] = (
    "Rating: " + dataset['rating'].astype(str) +
    ". Title: " + dataset['title'] +
    ". Review: " + dataset['review']
)

dataset
```

```python
clean_df = dataset.drop(['title', 'review', 'rating', 'net_helpfulness', 'helpfulness_count'], axis=1)

clean_df
```

```python
from sklearn.preprocessing import LabelEncoder

sentiment_encoder = LabelEncoder()
intent_encoder = LabelEncoder()

clean_df['sentiment_label'] = sentiment_encoder.fit_transform(clean_df['sentiment'])
clean_df['intent_label'] = intent_encoder.fit_transform(clean_df['intent'])
```

```python
min_samples = 2

valid_classes = clean_df["sentiment"].value_counts()
valid_classes = valid_classes[valid_classes >= min_samples].index

df = clean_df[clean_df["sentiment"].isin(valid_classes)]

df['sentiment'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df['sentiment_label']
)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

X_train = model.encode(train_df["text"].tolist(), show_progress_bar=True)
X_val = model.encode(val_df["text"].tolist(), show_progress_bar=True)
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_train_smoted, y_train_smoted = smote.fit_resample(X_train, train_df["sentiment_label"])
```

```python
from sklearn.linear_model import LogisticRegression

sentiment_model = LogisticRegression(max_iter=2000)

sentiment_model.fit(X_train, train_df["sentiment_label"])
```

```python
intent_model = LogisticRegression(max_iter=2000)

intent_model.fit(X_train, train_df["intent_label"])
```

```python
from sklearn.metrics import classification_report

sent_pred = sentiment_model.predict(X_val)

print(classification_report(val_df["sentiment_label"], sent_pred))
```

```python
def predict_review(text):
    embedding = model.encode([text])

    sentiment = sentiment_model.predict(embedding)
    intent = intent_model.predict(embedding)

    return (
        sentiment_encoder.inverse_transform(sentiment)[0],
        intent_encoder.inverse_transform(intent)[0]
    )
```

```python
predict_review("Rating: 5. Title: Honest Review. Review: UI is so messy I can't see anything on the main screen, but the app does accomplish what it advertised and made my life simpler.")
```
